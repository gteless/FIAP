Atualizar o último gráfico para refletir o seu estado, incluindo grid, eixos, etc.


```python
def plot_gastos_por_mes(dados, barra_destaque=None, cor_destaque='tab:green', figsize=(10, 6)):
    # Cria lista de cores padrão
    cores = ['tab:blue'] * len(dados)

    # Se o nome da barra a destacar existir no índice, muda a cor dela
    if barra_destaque in dados.index:
        idx = dados.index.get_loc(barra_destaque)
        cores[idx] = cor_destaque

    # Plota o gráfico de barras com as cores ajustadas
    axis = dados.plot(kind='bar', figsize=figsize, color=cores)

    # Ajusta rótulos e grade
    axis.set_ylabel("Gastos em Milhões de Reais")
    axis.set_xlabel("Meses")
    axis.grid(linestyle='--', linewidth=0.75, axis='both')

    return axis
```

```python
plot_gastos_por_mes(tabela_de_compracao, barra_destaque='35 São Paulo')
```

Colorir a cor do seu estado com uma cor diferente, colorir os estados maiores e menores com cores distintas


```python
def plot_gastos_por_mes(
    dados,
    seu_estado='23 Ceará',
    cor_seu_estado='tab:green',
    cor_maior='tab:red',
    cor_menor='tab:purple',
    cor_padrao='tab:blue',
    figsize=(12, 6)
):
    dados = dados.sort_values(ascending=False)

    # Verifica se o estado existe nos dados
    if seu_estado not in dados.index:
        raise ValueError(
            f"O estado '{seu_estado}' não está presente nos dados.")

    valor_seu_estado = dados[seu_estado]

    cores = []
    for estado in dados.index:
        if estado == seu_estado:
            cores.append(cor_seu_estado)
        elif dados[estado] > valor_seu_estado:
            cores.append(cor_maior)
        elif dados[estado] < valor_seu_estado:
            cores.append(cor_menor)
        else:
            cores.append(cor_padrao)  # caso empate

    # Plotando o gráfico
    axis = dados.plot(kind='bar', color=cores, figsize=figsize)

    axis.set_ylabel("Gastos em Milhões de Reais")
    axis.set_xlabel("Estados")
    axis.set_title("Gastos por Estado (comparado ao seu)")
    axis.set_xticklabels(axis.get_xticklabels(), rotation=90)
    axis.grid(linestyle='--', linewidth=0.7)

    return axis
```

```python
plot_gastos_por_mes(tabela_de_compracao, seu_estado='23 Ceará')
plt.show()
```

Gasto por população de dois estados. Escolher dois estados, plotar a comparação desses gastos de acordo com a população deles (Base IBGE por exemplo)


```python
def plot_gasto_per_capita(dados, estado1, estado2, pop_estados):
    # Verifica se os estados estão nos dados
    if estado1 not in dados.index or estado2 not in dados.index:
        raise ValueError("Um dos estados não está presente na base de dados.")

    # Coletando valores
    gasto1 = dados[estado1]
    gasto2 = dados[estado2]

    # Pegando população (em milhares, milhões, etc.)
    pop1 = pop_estados.get(estado1)
    pop2 = pop_estados.get(estado2)

    if pop1 is None or pop2 is None:
        raise ValueError("População de um dos estados não encontrada.")

    # Gasto per capita
    per_capita1 = gasto1 / pop1
    per_capita2 = gasto2 / pop2

    # Plot
    estados = [estado1, estado2]
    valores = [per_capita1, per_capita2]

    plt.figure(figsize=(8, 5))
    bars = plt.bar(estados, valores, color=['tab:orange', 'tab:cyan'])
    plt.ylabel("Gasto per capita (em milhões por pessoa)")
    plt.title("Comparativo de gasto per capita")
    plt.grid(axis='y', linestyle='--', linewidth=0.7)

    plt.show()
```

```python
populacoes = {
    '23 Ceará': 1.30,
    '35 São Paulo': 32
}

plot_gasto_per_capita(tabela_de_compracao, '23 Ceará',
                      '35 São Paulo', populacoes)
```

```python
def insere_gastos_e_gasto_por_habitante(ordenados_por_total, gastos_e_populacao, mes):
    gastos = ordenados_por_total[mes]
    gastos_e_populacao[f"gastos_{mes}"] = gastos * 1_000_000
    gastos_e_populacao[f"gasto_por_habitante_{mes}"] = gastos_e_populacao[f"gastos_{mes}"] / \
        gastos_e_populacao["populacao"]
    return gastos_e_populacao
```

```python
from datetime import date

meses = {
    "Jan": 1,
    "Fev": 2,
    "Mar": 3,
    "Abr": 4,
    "Mai": 5,
    "Jun": 6,
    "Jul": 7,
    "Ago": 8,
    "Set": 9,
    "Out": 10,
    "Nov": 11,
    "Dez": 12
}


def para_dia(ano_mes: str):
    # captura o ano pela string ano_mes, pegando os 4 primeiros caracteres
    ano: int = int(ano_mes[:4])
    # captura o ano pela string ano_mes, pegando os 5 últimos caracteres
    mes_como_string = ano_mes[5:]
    # transforma em int usando o dicionário
    mes: int = meses[mes_como_string]
    # retorna em uma lista o ano e mes capturados e o dia 1 como padrão
    return date(ano, mes, 1)
```

```python
mensal.index = mensal.index.map(para_dia)
mensal.head()
```

```python
estados = ["Amazonas", "Mato Grosso", "Ceará"]
filtrado = mensal_aberto.query("uf in @estados")

# Seleciona apenas colunas numéricas, exceto 'ano' e 'uf'
colunas_numericas = filtrado.select_dtypes(include="number").columns.difference(["ano"])

por_ano_dos_estados = (
    filtrado
    .groupby(["ano", "uf"])[colunas_numericas]
    .sum()
    .reset_index()
)
```
